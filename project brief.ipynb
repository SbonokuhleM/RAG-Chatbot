# RAG Research Assistant for Literature Review

Conducting a literature review is a time consuming process that requires researchers to search through large volumes of academic papers to identify relevant studies, understand existing research, compare methodologies, and locate supporting evidence. Traditional keyword-based search systems often return a large number of potentially relevant documents but do not provide sufficient contextual understanding of the information contained within them.

Large Language Models (LLMs) can assist with summarising and answering questions about academic literature, but they may generate inaccurate or unsupported information when they do not have direct access to the relevant research papers. This creates a risk of hallucinations and makes it difficult for researchers to verify the sources underlying generated responses.

This project proposes the development of a Retrieval-Augmented Generation (RAG) Research Assistant for Literature Reviews. The system will retrieve relevant passages from a collection of academic papers and provide them as context to a language model before generating a response. The system will also provide references to the retrieved sources, allowing for verification of the information presented.

# What is Retrieval-Augmented Generation (RAG)

Retrieval-Augmented Generation (RAG) is a way to make AI answers more reliable by combining searching for relevant information and then generating a response. Instead of guessing based only on old training data, it first finds useful data from external sources (like documents or databases) and then uses it to give a better answer.

1. Fetches up-to-date data and reduces incorrect or made up answers
2. Works well with specialized data like medical or legal content
3. No need to retrain the model every time new data comes in
4. Can use user-specific data to give more relevant responses
